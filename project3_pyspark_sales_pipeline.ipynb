```python
data = [
    (1, "Laptop", 2, 800),
    (2, "Phone", 5, 500),
    (3, "Headphones", 10, 50),
    (4, "Laptop", 1, 800),
    (5, "Phone", 3, 500)
]
columns=['order_id','products','quantity','unit_price']
df=spark.createDataFrame(data,columns)
df.show()
```

```python
df=df.withColumn(
    'total',
    df["quantity"]*df['unit_price']
)
df.show()

```

```python
df.filter(df['total']>1000).show()


```

### What is the total revenue for each product?

```python
df.groupBy("products").sum('total').show()

```

### Total quantity sold
### Average unit price

```python
from pyspark.sql.functions import sum,avg
df.groupBy('products').agg(sum("total")).show()
```

```python
from pyspark.sql.functions import when
df=df.withColumn(
    "price_category",
    when(df['unit_price']<100,'Budget')
    .when((df['unit_price']>=100) & (df['unit_price']<500),"Mid-range")
    .when(df['unit_price']>=500,"Premium")
)
df.show()
```

```python
from pyspark.sql.functions import col

df.filter(col('products').isNull()).show()
```


## Data Cleaning


```python
df_clean=df.dropDuplicates()
print("Orignal Rows :",df.count())
print("Duplicate Rows :",df_clean.count())
```
