```python
data = [
    (1001, "Laptop", "North", 2, 800),
    (1002, "Mobile", "South", 3, 400),
    (1003, "Keyboard", "East", 5, 50),
    (1004, "Monitor", "West", 2, 300),
    (1005, "Tablet", None, 1, 300),
    (1002, "Mobile", "South", 3, 400)
]

columns = [
    "order_id",
    "product",
    "region",
    "quantity",
    "unit_price"
]


orders_df=spark.createDataFrame(data,columns)
display(orders_df)
```

```python
# Inverstigation
orders_df.printSchema()
```

```python
# Total records
print("Total records:",orders_df.count())
```

```python
print(orders_df.groupBy("order_id").count().show())
```

```python
# now we will see numbers of non nulll values in each column
from pyspark.sql.functions import col,count

orders_df.select([
    count(col(c)).alias(c)
    for c in orders_df.columns
]).show()
```

```python
# identify exact duplicates records
# ye code har column ke individual duplicates nahi dhoondh raha. Ye code poori Row (sari columns ko mila kar) check kar raha hai ke koi row bilkul same hai ya nahi.
exact_duplicates=orders_df.groupBy(
    orders_df.columns
).count().filter(col("count")>1)
display(exact_duplicates)
```

```python
# duplicates orders
duplicate_orders=orders_df.groupBy("order_id").count().filter(col("count")>1)
display(duplicate_orders)

```

```python
from pyspark.sql import functions as F
# Counting Null values
null_counts=orders_df.select([
    F.sum(F.when(F.col(c).isNull(),1).otherwise(0).alias(c))
    for c in orders_df.columns
])
```
